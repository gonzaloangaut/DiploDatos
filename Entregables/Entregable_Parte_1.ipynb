### Conteo de frecuencias de los lenguajes de programación

La columna que contiene información sobre los lenguajes de programación utilizados es `tools_programming_languages`. Sus valores son strings con los lenguajes seleccionados separados por comas.



```python
df_filtered.tools_programming_languages[:3]
```


Las siguientes celdas de código separan estos lenguajes de programación y cuentan la frecuencia con la que aparecen.

```python
# Convert the comma-separated string of languages to a list of string.
# Remove 'ninguno de los anteriores' option, spaces and training commas.
def split_languages(languages_str):
  if not isinstance(languages_str, str):
    return []
  # Remove 'other' option
  languages_str = languages_str.lower()\
    .replace('ninguno de los anteriores', '')
  # Split string into list of items
  # Remove spaces and commas for each item
  return [lang.strip().replace(',', '')
          for lang in languages_str.split()]

# Create a new column with the list of languages
df_filtered.loc[:, 'cured_programming_languages'] = df_filtered.tools_programming_languages\
    .apply(split_languages)
if 'cured_programming_languages' not in relevant_columns:
    relevant_columns.append('cured_programming_languages')

# Duplicate each row of df for each programming language
# mentioned in the response.
# We only include in df_lang the columns we are going to analyze later, so we
# don't duplicate innecesary information.
df_lang = df_filtered.cured_programming_languages\
    .apply(pd.Series).stack()\
    .reset_index(level=-1, drop=True).to_frame()\
    .join(df_filtered[relevant_columns])\
    .rename(columns={0: 'programming_language'})
# Horrible programming style! But a lot of data science code can be written with
# as concatenations of functions (pipelines), and there's no elegant way of
# doing that on Python.
df_lang[:5]
```

En la columna `programming_language` se encuentra cada lenguaje por separado. Notar que si una respuesta contenía 3 lenguajes, como `"HTML, Javascript, Python"`, la fila ha sido replicada 3 veces. Por ello, hay tres filas con índice 1.

```python
language_count = df_lang.programming_language.value_counts()\
    .reset_index()\
    .rename(columns={'programming_language': 'language', 'count': 'frequency'})
    #.rename(columns={'index': 'language', 'programming_language': 'frequency'})
language_count[:10]
```

Ahora comencemos con el análisis de los datos. Como primera instancia, veremos un boxplot para saber de forma cualitativa cuánto gana la gente dependiendo en los lenguajes de programación.

```python
# Calcular la mediana del salario por lenguaje
median_salary_by_lang = filtered_df_lang.groupby('programming_language')['salary_monthly_NETO'].median()

# Ordenar los lenguajes por la mediana (de mayor a menor, podés usar ascending=False)
ordered_langs = median_salary_by_lang.sort_values(ascending=False).index

# Crear el boxplot ordenado
plt.figure(figsize=(12, 6))
seaborn.boxplot(
    data=filtered_df_lang,
    x='salary_monthly_NETO',
    y='programming_language',
    order=ordered_langs,
    hue='programming_language',
    palette='Set2'
)
plt.xlabel('Salario mensual NETO')
plt.ylabel('Lenguaje de programación')
plt.title('Boxplot del salario por lenguaje de programación (ordenado por mediana)')
plt.ticklabel_format(style='plain', axis='x')
plt.show()
```

También podemos graficar la media de sueldo neto para cada lenguaje

```python
# Agrupamos por lenguaje y calculamos la media de salario
mean_salary_by_lang = (
    filtered_df_lang
    .groupby('programming_language')['salary_monthly_NETO']
    .mean()
    .sort_values(ascending=False)
)
# Creamos el gráfico de barras
plt.figure(figsize=(12, 6))
seaborn.barplot(
    x=mean_salary_by_lang.values,
    y=mean_salary_by_lang.index,
    hue=mean_salary_by_lang.index,
    palette='Set3',
)
# Etiquetas
plt.xlabel('Salario mensual NETO (promedio)')
plt.ylabel('Lenguaje de programación')
plt.title('Salario promedio por lenguaje de programación')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

En estos gráficos podemos ver que los lenguajes de programación con media y mediana más altas son ruby, go y kotlin.

Veamos cual es la probabilidad de ganar por encima de la media (total) para cada lenguaje.

```python
# el salario medio entre todos los programadores (contando los lenguajes no relevantes) es
mean_salary = df_filtered['salary_monthly_NETO'].mean()
mean_salary
```

```python
# agrupamos por lenguaje de programación, para cada lenguaje toma la serie de 'salary_monthly_NETO'
# compara el sueldo con la media y al aplicar .mean() nos da la propoción de True
probs = filtered_df_lang.groupby('programming_language')['salary_monthly_NETO'].apply(
    lambda x: (x > mean_salary).mean()
)
# Luego probs nos da la probabilidad de ganar por encima de la media para cada lenguaje
probs.sort_values(ascending=False)
```

```python
# podemos obtener el mínimo y el máximo
lang_min = probs.idxmin()
prob_min = probs.min()
lang_max = probs.idxmax()
prob_max = probs.max()

print("El mínimo es: ", lang_min, " con: ", prob_min)
print("El máximo es: ", lang_max, " con: ", prob_max)
```

Por lo tanto, observamos que el lenguaje de programación de los relevantes con el que hay más probabilidades de ganar por encima de la media es kotlin, con un 71,54%. Por otro lado, el lenguaje con el que hay menos probabilidades de ganar por encima de la media es php, con un 35,38%.

Podemos ahora ver que lenguaje utiliza la gente que más plata gana. Para eso podemos ver cuanta gente gana por encima del percentil 90 dependiendo el lenguaje.

```python
# Lo hacemos de manera similar a lo de la media
percentil_90 = df_filtered['salary_monthly_NETO'].quantile(0.90)
percentil_90
```

```python
probs_90 = filtered_df_lang.groupby('programming_language')['salary_monthly_NETO'].apply(
    lambda x: (x > percentil_90).mean()
)
probs_90.sort_values(ascending=False)
```

```python
# podemos obtener el mínimo y el máximo
lang_min_90 = probs_90.idxmin()
prob_min_90 = probs_90.min()
lang_max_90 = probs_90.idxmax()
prob_max_90 = probs_90.max()

print("El mínimo es: ", lang_min_90, " con: ", prob_min_90)
print("El máximo es: ", lang_max_90, " con: ", prob_max_90)
```

Por lo tanto, observamos que el lenguaje de programación entre los relevantes con el que hay más probabilidades de ganar por encima del percentil 90 es ruby, con un 31,75%. Por otro lado, el lenguaje con el que hay menos probabilidades de ganar por encima del percentil 90 es .net, con un 7,90%.

Podemos hacer un análisis un poco más profundo y ver cuanto cambia el salario si la gente sabe un solo lenguaje de programación o más de uno.

```python
# Crear nuevo DataFrame que copia los lenguajes y el salario de df_filtered (que habia hecho el proceso
# de 'cured_programming_languages' pero no duplicó los datos como en df_lang)
df_number_lang = df_filtered[['cured_programming_languages', 'salary_monthly_NETO']].copy()
df_number_lang[:5]
```

```python
# Marcar si la persona sabe más de un lenguaje

# Función auxiliar para saber si hay más de un lenguaje
def knows_more_than_one(langs):
    return len(langs) > 1

# Aplicar la función
df_number_lang.loc[:, 'more_one_language'] = \
    df_number_lang['cured_programming_languages'].apply(knows_more_than_one)

# Calcular sueldos altos
df_number_lang.loc[:, 'higher_mean_salary'] = df_number_lang['salary_monthly_NETO'] > mean_salary

df_number_lang[:5]
```

```python
# Agrupamos y calculamos proporción
probs_by_lang_count = df_number_lang.groupby('more_one_language')['higher_mean_salary'].mean()
probs_by_lang_count
```

La probabilidad de ganar más que la media sabiendo 1 lenguaje es del 33,95% mientras que sabiendo más de uno es del 41,81%.

Para visualizar mejor estos datos, hagamos un boxplot y un gráfico de la media.

```python
plt.figure(figsize=(12, 6))
df_number_lang['more_one_language'] = df_number_lang['more_one_language'].astype('category')
seaborn.boxplot(data=df_number_lang,
                x='salary_monthly_NETO',
                y='more_one_language',
                hue='more_one_language',
                legend=False,
                palette='Set2')
plt.xlabel('Salario mensual NETO (promedio)')
plt.ylabel('¿Conoce más de un lenguaje?')
plt.title('Boxplot del salario según si se conoce más de un lenguaje')
plt.ticklabel_format(style='plain', axis='x')
```

```python
# Agrupamos por lenguaje y calculamos la media de salario
df_number_lang['more_one_language'] = df_number_lang['more_one_language'].astype('category')
mean_salary_by_lang = (
    df_number_lang
    .groupby('more_one_language', observed=False,)['salary_monthly_NETO']
    .mean()
    .sort_values(ascending=False)
)
# Creamos el gráfico de barras
plt.figure(figsize=(8, 6))
seaborn.barplot(
    x=mean_salary_by_lang.values,
    y=mean_salary_by_lang.index,
    hue=mean_salary_by_lang.index,
    palette='Set2',
    legend=False,
)
# Etiquetas
plt.xlabel('Salario mensual NETO (promedio)')
plt.ylabel('¿Conoce más de un lenguaje?')
plt.title('Salario según si se conoce más de un lenguaje')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

En conclusión, los lenguajes de programación que muestran mayor utilidad en términos salariales son kotlin y ruby. Ambos presentan las medias salariales más elevadas. Por un lado, kotlin es el lenguaje que ofrece la mayor probabilidad de ganar por encima del salario medio y la mediana más alta. Por otro lado, ruby es el lenguaje más común entre quienes se encuentran por encima del percentil 90 de ingreso.

También observamos que php y .net no parecen ser lenguajes particularmente valorado en el mercado, al menos en términos de salario. Son los que presentan la media y mediana más bajas entre los diez lenguajes más utilizados y la menor probabilidad de alcanzar o superar la media salarial.

Por último, nuestros datos indican que conocer más de un lenguaje de programación se asocia con mayores ingresos.

Vemos que las dos variables representan una relación lineal por lo que podemos decir que están correlacionadas. Para poder tener mayor seguridad en eso, podemos estudiar el coeficiente de Pearson que es una medida de dependencia lineal entre dos variables aleatorias cuantitativas.

Este coeficiente es alto, por lo que existe una correlación muy fuerte entre el sueldo bruto y el sueldo neto. Esto sugiere que una variable puede predecir bastante bien a la otra. Por lo tanto, se podría eliminar la variable de salario bruto de la encuesta.

```python
# Filtrar las dos subpoblaciones
universitarios = df_filtered_2[df_filtered_2['profile_studies_level'] == 'Universitario']
terciarios = df_filtered_2[df_filtered_2['profile_studies_level'] == 'Terciario']

# Crear figura con dos subplots
fig, axes = plt.subplots(ncols=2, figsize=(15,6))

# Histograma con frecuencias absolutas (count)
seaborn.histplot(universitarios['salary_monthly_NETO'], color='green', label='Universitario', stat="count", bins=50, ax=axes[0])
seaborn.histplot(terciarios['salary_monthly_NETO'], color='blue', label='Terciario', stat="count", bins=50, ax=axes[0])
axes[0].set_title("Frecuencia absoluta")
axes[0].set_xlabel("Salario mensual neto")
axes[0].set_ylabel("Cantidad de personas")
axes[0].legend()

# Histograma con densidad (normalizado)
seaborn.histplot(universitarios['salary_monthly_NETO'], color='green', label='Universitario', stat="density", bins=50, ax=axes[1])
seaborn.histplot(terciarios['salary_monthly_NETO'], color='blue', label='Terciario', stat="density", bins=50, ax=axes[1])
axes[1].set_title("Distribución normalizada")
axes[1].set_xlabel("Salario mensual neto")
axes[1].set_ylabel("Densidad")
axes[1].legend()

plt.suptitle("Distribución del salario según nivel de estudio", fontsize=16)
plt.tight_layout()
plt.show()
```

Calculemos ahora las medidas de centralización y dispersión para cada distribución.

```python
# Media y mediana
print("Universitarios:")
print("  Media:", universitarios['salary_monthly_NETO'].mean())
print("  Mediana:", universitarios['salary_monthly_NETO'].median())
print("  Desviación estándar:", universitarios['salary_monthly_NETO'].std())
print()

print("Terciarios:")
print("  Media:", terciarios['salary_monthly_NETO'].mean())
print("  Mediana:", terciarios['salary_monthly_NETO'].median())
print("  Desviación estándar:", terciarios['salary_monthly_NETO'].std())
```

Haciendo un análisis de los valores obtenidos, vemos que los universitarios, en promedio, ganan más que los terciarios, esto se refleja tanto en la media como en la mediana. Sin embargo, la desviación estándar también es más grande, lo que nos muestra que hay más heterogeneidad en los sueldos de los universitarios. La diferencia entre la media y la mediana en ambos grupos sugiere distribuciones asimétricas lo cual también se ve en los histogramas.

El nivel de estudio no parece ser independiente del salario mensual neto, ya que existen diferencias claras en las distribuciones y medidas centrales, pero las formas de las distribuciones son similares.
