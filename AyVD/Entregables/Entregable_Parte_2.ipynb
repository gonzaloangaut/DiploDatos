## Ejercicio 1: Estimación

**Consigna:**  Calcular una estimación puntual y un intervalo de confianza de nivel (1-alpha) para la resta entre la media del salario Neto para Hombres y la media del salario Neto para mujeres (diferencia de las medias entre el grupoA y grupoB).
¿Cómo se relaciona este intervalo de confianza con el test de hipótesis para diferencia de medias?

```python
import numpy as np
from scipy import stats

# Estimador puntual
estimador_puntual = groupA.mean() - groupB.mean()
print(f"Estimador puntual: {estimador_puntual:.2f}")

# alpha
alpha = 0.05

# Tamaños muestrales
nA = len(groupA)
nB = len(groupB)

# Desvíos estándar muestrales
sA = np.std(groupA, ddof=1)
sB = np.std(groupB, ddof=1)

# Error estándar de la diferencia de medias
se = np.sqrt(sA**2 / nA + sB**2 / nB)

# Grados de libertad aproximados usando la fórmula de Welch-Satterthwaite
df = (sA**2 / nA + sB**2 / nB)**2 / ((sA**2 / nA)**2 / (nA-1) + (sB**2 / nB)**2 / (nB-1))

# Valor crítico (t de Student, pues no conocemos la desviación estándat)
t_crit = stats.t.ppf(1 - alpha/2, df)

# Margen de error
margen_error = t_crit * se

# Intervalo de confianza
ic_inferior = estimador_puntual - margen_error
ic_superior = estimador_puntual + margen_error
longitud_ic = ic_superior - ic_inferior

print(f"Intervalo de confianza: ({ic_inferior:.2f}, {ic_superior:.2f})")
print(f"Longitud del intervalo: {longitud_ic:.2f}")
```

Este intervalo de confianza se relaciona con el test de hipótesis porque nos permite evaluar si la diferencia entre las medias es estadísticamente significativa.

Podemos plantear la hipótesis nula $H_0$ como que la diferencia de medias es cero ($\mu_A - \mu_B = 0$), y la hipótesis alternativa $H_1$ como la diferencia de medias es distinta de cero ($\mu_A - \mu_B \neq 0$). 

Si el intervalo de confianza contiene al $0$, entonces no podemos rechazar la hipótesis nula (no encontramos evidencia suficiente de una diferencia significativa).

Si el intervalo de confianza no contiene al $0$, entonces rechazamos la hipótesis nula y aceptamos la hipótesis alternativa (existe una diferencia significativa).

En nuestro caso, el intervalo de confianza no contiene al $0$, por lo que rechazamos la hipótesis nula. Esto nos permite concluir que existe una diferencia significativa entre las medias de salario de hombres y mujeres.

## Ejercicio 2: Test de hipótesis



### 2.1 Formalización

Describir formalmente los distintos componentes de un test de hipótesis para comprobar si la distribución de los salarios es distinta entre los grupos A y B.

**Hipótesis Nula**

$H_0=$ La diferencia de medias es cero ($\mu_A - \mu_B = 0$).

**Hipótesis Alternativa**

$H_1=$ La diferencia de medias es distinta de cero ($\mu_A - \mu_B \neq 0$).

**Estadístico (Pivote)**

El estadístico que utilizamos es el $T$ ya que no conocemos las varianzas poblacionales y usamos las varianzas muestrales. Este se define para una sola muestra como:

$T=\frac{\bar{X}-\mu}{S/\sqrt{n}}$.

En nuestro caso particular, como vemos la diferencia entre medias, tenemos 2 muestras y esta ecuación se convierte en:

$T=\frac{(\bar{X}_A-\bar{X}_B)-0}{\sqrt{\frac{S_A^2}{n_A}+\frac{S_B^2}{n_B}}}$,

donde:
- $\bar{X}_A$ es la media muestral de A.
- $\bar{X}_B$ es la media muestral de B.
- $S_A^2$ es la varianza muestral de A.
- $S_B^2$ es la varianza muestral de B.
- $n_A$, $n_B$ es el tamaño muestral de cada grupo.

**Distribución bajo $H_0$**

Bajo $H_0$, el estadístico $T$ sigue una distribución **t de Student** con **grados de libertad aproximados** dados por la fórmula de Welch-Satterthwaite:

$
df = \frac{\left( \frac{s_A^2}{n_A} + \frac{s_B^2}{n_B} \right)^2}{\frac{\left( \frac{s_A^2}{n_A} \right)^2}{n_A-1} + \frac{\left( \frac{s_B^2}{n_B} \right)^2}{n_B-1}}
$.

O sea, $T \sim t_{df}$ bajo $H_0$.

### 2.2 P-valor

1. Calcule el p-valor y decida si rechazar o no la hipótesis nula.
2. Interprete el resultado.

Links útiles:
* [Test de hipótesis usando scipy](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.ttest_ind.html)
* [Test de Welch](http://daniellakens.blogspot.com/2015/01/always-use-welchs-t-test-instead-of.html)

Vamos a calcular el **p-valor** para los 2 grupos usando el test de Welch. Para eso, usamos stats.ttest_ind() y agregamos, para usar Welch, equal_var = False. Luego, el p-valor se devuelve en la posición 1. Si el p-valor es menor que $\alpha$, rechazamos la hipótesis nula y aceptamos la alternativa, mientras que si es mayor o igual, no tenemos suficiente evidencia para rechazarla.

```python
# Calculemos el p-valor usando Welch
p_valor = stats.ttest_ind(groupA , groupB, equal_var = False)[1]
print(f"El p-valor es: {p_valor}")

# Decisión
if p_valor < alpha:
    # Si el p-valor es menor que alpha, entonces rechazamos la hipótesis nula
    print("Rechazamos la hipótesis nula pues hay evidencia suficiente para hacerlo.")
else:
    # Si no, no podemos rechazarla
    print("No podemos rechazar la hipótesis nula debido a la falta de evidencia.")
```

El test de hipótesis nos indica que podemos rechazar la hipótesis nula. Por lo tanto, tenemos evidencia estadística para afirmar que la diferencia de medias es distinta de cero ($\mu_A - \mu_B \neq 0$), es decir, que uno de los grupos gana, en promedio, más que el otro.

### [Opcional] 2.3 Potencia del test

Nuestra muestra, ¿era lo suficientemente grande para detectar si existe o no una diferencia entre los grupos?

1. Utilice la función `tt_ind_solve_power` para calcular el tamaño necesario de la muestra para un poder estadístico de 0.8, 0.9 y 0.95, asumiendo una significancia estadística de 0.05.
2. ¿Cómo intepretan el poder estadístico de un test? Dado su conocimiento de dominio sobre los datos, ¿les parece que esta muestra es lo suficientemente grande para ser representativo de la tendencia general? ¿y para utilizarlo en un juicio penal contra una empresa XX por una causa de discriminación?

[Documentación](https://www.statsmodels.org/stable/generated/statsmodels.stats.power.tt_ind_solve_power.html)

NOTA: este análisis debería hacerse ANTES de recolectar los datos.

```python
from statsmodels.stats.power import tt_ind_solve_power
```

```python
effect_size = (groupA.mean() - groupB.mean()) / groupB.std()
# nobs1=None  - What we want to know
alpha = 0.05
ratio = len(groupB) / len(groupA) # Al hacer esta elección, nobs1 es el tamaño del grupo A
powers = [0.8, 0.9, 0.95]

for power in powers:
    length = tt_ind_solve_power(effect_size=effect_size, alpha=alpha,power=power,  ratio=ratio)
    if len(groupA)>= length:
        print(f"\nEl tamaño necesario de la muestra del grupo A para un poder estadístico de {power} es {round(length,2)}.\n"
              f"Como el tamaño de la muestra del grupo A es {len(groupA)}, podemos afirmar que la muestra\n"
              f"era lo suficientemente grande para un poder estadístico de {power}.")

    else:
        print(f"\nEl tamaño necesario de la muestra del grupo A para un poder estadístico de {power} es {round(length,2)}.\n"
              f"Como el tamaño de la muestra del grupo A es {len(groupA)}, podemos decir que la muestra no\n"
              f"era lo suficientemente grande para un poder estadístico de {power}.")   
```

El poder estadístico de un test es la probabilidad de que es test rechace correctamente la hipótesis nula si la hipótesis alternativa es verdadera. Es decir, si un test tiene alto poder estadístico, entonces tiene bajas probabilidades de cometer un error de tipo II (un falso negativo).

En particular, estas muestras parecen ser lo suficientemente grandes como para ser representativa de la tendencia general. Sin embargo, habría que tener precauciones obre cómo se obtuvieron los datos ya que, para ser representativa de una población, necesitamos, además de que sea grande, que no esté sesgada.
