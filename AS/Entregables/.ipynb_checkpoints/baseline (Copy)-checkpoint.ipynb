### Distribución de datos

Para las columnas categóricas y las numéricas binarias (como `hypertension` que solo puede tomar valores 0 y 1) veamos que no hay valores mal catalogados:

```python
# Vemos primero las binarias
binary_cols = [
    'hypertension',
    'heart_disease',
]
for col in binary_cols:
    print(f"Valores posibles para {col}: ", train_df[col].unique())

# y las categóricas
categorical_cols = [
    'gender',
    'smoking_history',
]
for col in categorical_cols:
    print(f"Valores posibles para {col}: ", train_df[col].unique())
```

Podemos observar que todos los resultados posibles son consistentes.

Para el resto de las variables numéricas vamos a ver su distribución.

```python
numeric_cols = [
  'age',
  'bmi',
  'HbA1c_level',
  'blood_glucose_level',
]

# Veamos la distribución de las columnas numéricas relevantes
for col in numeric_cols:
    sns.histplot(train_df[col].dropna(), bins=30)
    plt.title(f'Distribución de {col}')
    plt.show()
```

Si eliminamos outliers:

```python
train_df_filtered = train_df.copy()

# Armamos una máscara donde guardamos a los valores outliers
outlier_mask = np.full(len(train_df_filtered), False)
for col in numeric_cols:
    # Podemos eliminar los datos por encima del percentil 99 y debajo del percentil 1
    lower = train_df_filtered[col].quantile(0.01) # Podemos agrandar este rango si asi lo quieren
    upper = train_df_filtered[col].quantile(0.99)

    # Identificamos outliers por columna
    is_outlier = (train_df_filtered[col] < lower) | (train_df_filtered[col] > upper)

    # Acumulamos una máscara global de outliers
    outlier_mask |= is_outlier

# Filtramos todas las filas que tienen outliers en alguna columna
train_df_filtered = train_df_filtered[~outlier_mask]
```

```python
# Veamos la nueva distribución de las columnas numéricas relevantes
for col in numeric_cols:
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))
    fig.suptitle(f'Distribución de {col}')

    # Antes de eliminar outliers
    sns.histplot(train_df[col].dropna(), bins=30, ax=axes[0], color='skyblue', edgecolor='black')
    axes[0].set_title('Antes de eliminar outliers')

    # Después
    sns.histplot(train_df_filtered[col].dropna(), bins=30, ax=axes[1], color='lightgreen', edgecolor='black')
    axes[1].set_title('Después de eliminar outliers')
    plt.show()
```

Notemos que no se observan outliers que alteren significativamente las distribuciones.

### Imputación de datos

Necesitamos contruir nuestros conjuntos de datos para realizar los entrenamientos:

Comencemos viendo la cantidad de valores faltantes que tiene cada columna.

```python
train_df_filtered.isnull().sum()
```

¡Ninguna tiene datos faltantes! Por lo tanto, no hay datos que imputar.

Sin embargo, notemos que:

```python
train_df_filtered[train_df_filtered['gender']=='Other']
```

Es decir, los casos en la base de datos con `gender`='Other' corresponden a usuarios sin diabetes. Esto induce una relación inexistente:

- Si `gender`='Other', entonces `diabetes` = 0

Esta regla no representa a la realidad y se debe a:

- Muy baja frecuencia de esta categoría (solo unas pocas filas).

- Ausencia total de casos positivos en dicha categoría.

Mantener esta variable tal cual podría generar un modelo con lógica incorrecta que dependa de una coincidencia circunstancial del dataset. Modelos probabilísticos como Naive Bayes aprenden la regla trivial anterior y obtienen métricas cercanas al 100% sin realmente generalizar.
Por lo tanto, decidimos eliminar estos registros para evitar introducir sesgos y métricas artificialmente altas.

```python
#train_df_filtered = train_df_filtered[train_df_filtered['gender'] != 'Other']
train_df_filtered
```

Separamos la base de datos de modo que la variable `X` contenga las columnas utilizadas como predictores, y `y` la variable objetivo a predecir.

```python
y = train_df_filtered.diabetes
X = train_df_filtered.drop('diabetes',axis=1)
X
```

Vamos a separar la columna con los Id de los pacientes en una sola variable:

```python
patientId = X.patient
X.drop('patient',axis=1,inplace=True)
X
```

### Encoder

Necesitamos transformar las variables que son categoricas a númericas, como por ejemplo **gender** y **smoking_history**

Para esto, vamos a usar *OneHotEncoder*.

```python
categorical_cols = ['gender', 'smoking_history']

encoder = OneHotEncoder(drop='first', sparse_output=False)
X_cat_encoded = encoder.fit_transform(X[categorical_cols])

# Obtener nombres de columnas codificadas
encoded_cols = encoder.get_feature_names_out(categorical_cols)

# Convertir a DataFrame
X_cat_encoded_df = pd.DataFrame(X_cat_encoded, columns=encoded_cols, index=X.index)

X_encoded = pd.concat([X.drop(columns=categorical_cols), X_cat_encoded_df], axis=1)
X_encoded
```

```python
#print(X.columns)
#label_cols = ["gender", "smoking_history"]

#for col in label_cols:
#    X[col] = X[col].astype(str)
#    X[col] = LabelEncoder().fit_transform(X[col])
#X
```

### Escaleo

Necesitamos también estandarizar las variables:

```python
x_names = X_encoded.columns
X_encoded = StandardScaler().fit_transform(X_encoded)
X_encoded = pd.DataFrame(X_encoded, columns=x_names)
X_encoded
```

A partir de los datos en las variables `X` e `y` separamos en train y en test:

```python
x_train, x_test, y_train, y_test = train_test_split(X_encoded, y, train_size=0.8, random_state = 8)
```

### Planteo de modelos
Tienen que elegir los 2 mejores modelos así que prueben con varios.

```python
clfs =  [DecisionTreeClassifier(),
        RandomForestClassifier(),
        LogisticRegression(),
        MLPClassifier(),
        XGBClassifier(),
        GaussianNB(),
        LinearSVC()]

names = ['Arbol de decisión',
        'Random Forest',
        'Regresión Logística',
        'Perceptrón multicapa',
        'XGBoost',
        'Naive Bayes',
        'SVM']

trained_models = []
accuracy_models = []
for clf,name in zip(clfs,names):
    print(name)
    clf.fit(x_train, y_train)
    train_predictions = clf.predict(x_train)
    recall = recall_score(y_train, train_predictions)
    print(f"Recall train {name}: %.2f%%" % (recall * 100.0))

    train_predictions = clf.predict(x_test)
    recall = recall_score(y_test, train_predictions)
    print(f"Recall test {name}: %.2f%%" % (recall * 100.0))
    # plot_confusion_matrix(clf,x_test,y_test)
    trained_models.append(clf)
    accuracy_models.append(recall*100)
```

Observamos que el Arbol de decisión y el Random Forest predicen muy bien a los datos de entrenamiento pero no generalizan de la mejor manera. El segundo modelo que mejor predice los datos de testeo es el `XGBoost`, y tiene un gap menor entre los resultados de los datos de entrenamiento y de testeo. Vamos a usar este modelo.

### Algún tipo de ajuste de hiperparámetros de los modelos elegidos

```python
#grid = GridSearchCV(DecisionTreeClassifier(),
# {'criterion':('gini','entropy'),
# 'splitter':("best","random"),
#  'min_samples_leaf':(1, 2, 5),
#  'min_samples_split':(2, 3, 5, 10, 50, 100),
# 'max_depth':(5,10,20)})
#grid.fit(x_train,y_train)
#grid.best_params_
```

```python
# Modelo base
dt_model = DecisionTreeClassifier(random_state=42, class_weight='balanced')

# Grilla de hiperparámetros
param_grid_dt = {
    'criterion': ['gini', 'entropy'],
    'splitter': ['best', 'random'],
    'max_depth': [3, 5, 10, 20, None],
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [1, 2, 5, 10]
}

grid_dt = GridSearchCV(
    estimator=dt_model,
    param_grid=param_grid_dt,
    scoring='recall',
    cv=5,
    n_jobs=-1
)

grid_dt.fit(x_train, y_train)

print("Best parameters (Decision Tree):", grid_dt.best_params_)
print("Best recall (Decision Tree):", grid_dt.best_score_)
```

```python
clf = DecisionTreeClassifier(**grid_dt.best_params_).fit(x_train, y_train)

train_predictions_dt = clf.predict(x_train)
recall_dt = recall_score(y_train, train_predictions_dt)
print("Recall train Tree: %.2f%%" % (recall_dt * 100.0))

test_predictions_dt = clf.predict(x_test)
recall_dt = recall_score(y_test, test_predictions_dt)
print("Recall test Tree: %.2f%%" % (recall_dt * 100.0))
ConfusionMatrixDisplay.from_predictions(y_test, test_predictions_dt)
plt.show()

print(classification_report(y_test,clf.predict(x_test)))
```

```python
from collections import Counter

# Calcular peso positivo
counter = Counter(y_train)
neg, pos = counter[0], counter[1]
scale_pos_weight = neg / pos

xgb_model = XGBClassifier(
    eval_metric='logloss',
    random_state=42,
    scale_pos_weight=scale_pos_weight
)

param_grid_xgb = {
    'n_estimators': [100, 200, 400],
    'max_depth': [5, 7, 9],
    'learning_rate': [0.05, 0.1],
    'subsample': [0.8, 1],
    'colsample_bytree': [0.6, 0.8, 1],
    'min_child_weight': [1, 3, 5],
    'gamma': [0, 0.5, 1]
}

grid_xgb = GridSearchCV(
    xgb_model,
    param_grid_xgb,
    scoring='recall',
    cv=5,
    n_jobs=-1
)

grid_xgb.fit(x_train, y_train)

print("Best parameters:", grid_xgb.best_params_)
print("Best recall (CV):", grid_xgb.best_score_)
```

```python
clf = XGBClassifier(**grid_xgb.best_params_).fit(x_train, y_train)

train_predictions_xgb = clf.predict(x_train)
recall_xgb = recall_score(y_train, train_predictions_xgb)
print("Recall train XGBoost: %.2f%%" % (recall_xgb * 100.0))

test_predictions_xgb = clf.predict(x_test)
recall_xgb = recall_score(y_test, test_predictions_xgb)
print("Recall test XGBoost: %.2f%%" % (recall * 100.0))
ConfusionMatrixDisplay.from_predictions(y_test, test_predictions_xgb)
plt.show()

print(classification_report(y_test,clf.predict(x_test)))
```

```python
# Definimos el modelo base
rf_model = RandomForestClassifier(random_state=42, class_weight='balanced')

param_grid_rf = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 5, 10, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 5]
}

grid_rf = GridSearchCV(
    rf_model, 
    param_grid_rf, 
    scoring='recall',
    cv=5,
    n_jobs=-1
)
grid_rf.fit(x_train, y_train)

print("Best params:", grid_rf.best_params_)
print("Best recall:", grid_rf.best_score_)
```

```python
clf = RandomForestClassifier(**grid_rf.best_params_).fit(x_train, y_train)

train_predictions_rf = clf.predict(x_train)
recall_rf = recall_score(y_train, train_predictions_rf)
print("Recall train Tree: %.2f%%" % (recall_rf * 100.0))

test_predictions_rf = clf.predict(x_test)
recall_rf = recall_score(y_test, test_predictions_rf)
print("Recall test Tree: %.2f%%" % (recall_rf * 100.0))
ConfusionMatrixDisplay.from_predictions(y_test, test_predictions_rf)
plt.show()

print(classification_report(y_test,clf.predict(x_test)))
```

```python
clf = GaussianNB().fit(x_train, y_train)

train_predictions_rf = clf.predict(x_train)
recall_rf = recall_score(y_train, train_predictions_rf)
print("Recall train Tree: %.2f%%" % (recall_rf * 100.0))

test_predictions_rf = clf.predict(x_test)
recall_rf = recall_score(y_test, test_predictions_rf)
print("Recall test Tree: %.2f%%" % (recall_rf * 100.0))
ConfusionMatrixDisplay.from_predictions(y_test, test_predictions_rf)
plt.show()

print(classification_report(y_test,clf.predict(x_test)))
```
